```python
from transformers import pipeline
import torch

print("GPU available:", torch.cuda.is_available())
print("GPU name:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "No GPU")

prompt_1 = "List the top three benefits of AI in healthcare."

models = {
    "GPT-Neo 125M": "EleutherAI/gpt-neo-125M",
    "GPT-Neo 1.3B": "EleutherAI/gpt-neo-1.3B",
}

generators = {}

for name, model_path in models.items():
    generators[name] = pipeline("text-generation", model=model_path, device=0)

responses = {}

for name, generator in generators.items():
    output = generator(prompt_1, max_length=100, do_sample=True, temperature=0.7)
    responses[name] = output[0]["generated_text"]
    print(f"\n=== {name} Output ===")
    print(responses[name])

print("\n========================")
print("SUMMARY OF ALL RESPONSES")
print("========================\n")

for name, text in responses.items():
    print(f"--- {name} ---")
    print(text)
    print("\n")
```

Analysis

The experiment shows that model size has a noticeable impact on the quality of the generated responses. The GPT-Neo 125M model produced text that was repetitive, unfocused, and included irrelevant phrases such as “Image credit: Google,” which suggests limited understanding of the prompt. In comparison, the GPT-Neo 1.3B model gave a more coherent and organized response, although it still wasn’t completely accurate. Overall, the larger model demonstrated better clarity, stronger context handling, and fewer random or irrelevant outputs.

When comparing performance, the 1.3B model performed better. Its response was more structured and showed a clearer attempt to address the question, despite repeating some lines. The 125M model struggled significantly, often repeating entire sentences and failing to list the actual benefits requested by the prompt. This highlights how smaller models often lack the capacity to follow instructions effectively.

Both models also showed certain limitations. The 125M model was highly repetitive, introduced irrelevant content, and didn’t provide meaningful answers. The 1.3B model was an improvement but still repeated phrases and missed the expected top three benefits. Overall, both models had difficulty with instruction-following and tended to either be too verbose or too vague. These observations emphasize the limitations of older GPT-Neo models compared to more modern language models.

1. Prompt

Technique used:

-Zero-shot prompting: The model is not given any examples of how to answer. It must figure it out on its own.

-Context-based prompting: The model is given extra information to analyze — in this case, the product review text.

2. Prompt Evaluation

Context: Good, the review is included.

Precision: Moderate, asks for satisfaction + problems but not specific details.

Clarity: Clear, but could be more direct.

Sensitivity: Safe, no sensitive content.

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="EleutherAI/gpt-neo-1.3B",
    device=0
)

review = "The product has a sleek design and functions well for basic tasks. However, the battery drains quickly, which makes it hard to rely on during travel. Overall, it's a decent product, but the battery issue is a drawback."

# -------- ORIGINAL PROMPT --------
original_prompt = (
    "Read the following product review and explain how the user's experience "
    "with the product might influence their opinion about recommending it to others. "
    "Include aspects of their satisfaction and any problems they faced. "
    "Review: " + review
)

response_original = generator(
    original_prompt,
    max_length=len(original_prompt.split()) + 40,
    do_sample=False,
    num_beams=1,
    temperature=0,
    truncation=True
)

print("\n=== ORIGINAL PROMPT OUTPUT ===\n")
print(response_original[0]["generated_text"])


# -------- STRICT IMPROVED PROMPT --------
strict_prompt = (
    "ONLY use the exact words found in the review. "
    "Do NOT add any new information. "
    "Write 2–3 sentences summarizing the positive experience and the negative experience. "
    "Review: " + review
)

response_strict = generator(
    strict_prompt,
    max_length=len(strict_prompt.split()) + 40,
    do_sample=False,
    num_beams=1,
    temperature=0,
    truncation=True
)

print("\n=== IMPROVED PROMPT OUTPUT ===\n")
print(response_strict[0]["generated_text"])
```

Context

The prompt includes the product review, so the model has enough information to analyze. However, the outputs show that the model often ignored the review context and repeated the same sentences, meaning the context was provided but not well utilized.

Precision

Precision was not strong. The model was instructed to “explain how the user’s experience influences their recommendation,” but instead it repeated the review text multiple times. This shows the prompt lacks precise constraints, causing the model to drift or echo content.

Clarity

The prompt is clear, but the model’s responses show that clarity alone wasn’t enough. Without clearer structural instructions (e.g., “summarize in 3 sentences”), the model did not produce analytical output. This is why it repeatedly output the same lines.

```python
from transformers import pipeline

# Load model
generator = pipeline("text-generation", model="EleutherAI/gpt-neo-1.3B", device=0)

# -----------------------------
# ORIGINAL TEMPLATE FORM PROMPT
# -----------------------------
original_prompt = (
    "Please fill out the feedback form below:\n\n"
    "Product Name: [Enter product name here]\n"
    "Feedback Type (Positive / Negative / Neutral): [Select one]\n"
    "Comments (Max 40 words): [Enter your feedback here]\n"
)

response_original = generator(
    original_prompt,
    max_length=120,
    do_sample=True
)

print("=== ORIGINAL PROMPT OUTPUT ===\n")
print(response_original[0]["generated_text"])


# -----------------------------
# IMPROVED TEMPLATE FORM PROMPT
# -----------------------------
improved_prompt = (
    "You are filling out a structured product feedback form.\n"
    "Provide short, clear answers.\n\n"
    "Product Name: [Enter product name]\n"
    "Feedback Type (Positive / Negative / Neutral): [Choose one]\n"
    "Comments (Max 40 words, do NOT exceed the limit): [Write brief feedback here]\n"
)

response_improved = generator(
    improved_prompt,
    max_length=120,
    do_sample=True
)

print("\n=== IMPROVED PROMPT OUTPUT ===\n")
print(response_improved[0]["generated_text"])
```

Testing the original prompt showed that the model added many unrelated fields and did not follow the form. After improving the prompt with clearer instructions, the output became more structured, but the model still drifted into irrelevant text. The improved prompt increased clarity and reduced randomness slightly, but GPT-Neo continued to hallucinate because it is not instruction-tuned and does not reliably follow strict formatting.
